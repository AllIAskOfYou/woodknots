```python
import numpy as np
import json
from mltools.matching.utils import (
    parse_coco_annotations,
    construct_voxel_grid,
    correct_annotations,
    confusion_matrix,
)
```

```python
ann_file = "../../data/matching/annotations/instances_default.json"
projections_dir = "../../data/matching/resized_1024_3d_v3_distortion"
corrections_file = "../../data/matching/annotations/corrections.json"
voxel_size = 0.005
annotations = parse_coco_annotations(ann_file)
with open(corrections_file, "r") as f:
    corrections = json.load(f)
correct_annotations(annotations, corrections)
vg = construct_voxel_grid(annotations, projections_dir, voxel_size)
```

```python
structures = vg.find_structures(merge_threshold = 0.2)
tp, fp, fn, tn = confusion_matrix(annotations, structures)
print(f"True Positives: {tp}\nFalse Positives: {fp}\nTrue Negatives: {tn}\nFalse Negatives: {fn}")
precision = tp / (tp + fp) if (tp + fp) > 0 else 0
recall = tp / (tp + fn) if (tp + fn) > 0 else 0
print(f"Precision: {precision:.4f}, Recall: {recall:.4f}")
```
