```python
import numpy as np
from matplotlib import pyplot as plt

def radialk3(points, ks):
    k1, k2, k3 = ks
    r2 = points[:, 0] * points[:, 0] + points[:, 1] * points[:, 1]
    r4 = r2 * r2
    r6 = r2 * r4
    factor = 1 + k1 * r2 + k2 * r4 + k3 * r6
    distorted = points * factor[:, np.newaxis]
    return distorted
```

```python

ks = [-0.18256676708360231, 0.1730017203231568, -0.041831834873593855]
points = np.meshgrid(np.linspace(-1, 1, 21), np.linspace(-1, 1, 21))
points = np.stack(points, axis=-1).reshape(-1, 2)
distorted = radialk3(points, ks)

# draw lines from original points to corresponding distorted points
plt.figure(figsize=(5,5))
# plt.scatter(points[:, 0], points[:, 1], s=1, color='blue', label='Original Points')
for i in range(0, len(points), 1):
    plt.plot([points[i, 0], distorted[i, 0]], [points[i, 1], distorted[i, 1]], 'k-', alpha=0.5)
plt.show()
```
