```python
from pathlib import Path
import json

groups_file = "/home/matic/mltools/metadata/groups.json"
with open(groups_file, "r") as f:
    groups = json.load(f)

data_dir = Path("/mnt/d/matics/logs/logs_600")
# read all image names from the data directory
image_dir = data_dir / "images"
image_names = [f.stem for f in image_dir.glob("*")]
suffixes = set([f.suffix for f in image_dir.glob("*")])
print(f"Found {len(image_names)} images with suffixes: {suffixes}")

# remove all that are not in image_names from the groups
filtered_groups = []
for group in groups:
    # print(f"Before: {len(group)}")
    filtered_group = [name for name in group if name in image_names]
    filtered_groups.append(filtered_group)
    # print(f"After: {len(filtered_group)}")

total_filtered = sum(len(group) for group in filtered_groups)
print(f"Total filtered groups: {total_filtered}")

print(filtered_groups)
new_groups_file = "/mnt/d/matics/logs/logs_600/groups.json"
with open(new_groups_file, "w") as f:
    json.dump(filtered_groups, f, indent=4)
```

```python
# rename all images into .jpg if they are .JPG
for image_file in image_dir.glob("*.JPG"):
    new_file = image_file.with_suffix(".jpg")
    image_file.rename(new_file)
```

```python
annotations_file = data_dir / "annotations.json"

with open(annotations_file, "r") as f:
    annotations = json.load(f)
```

```python
for annotation in annotations["annotations"]:
    attributes = annotation["attributes"]
    for key in ["occluded", "type"]:
        if key in attributes:
            del attributes[key]
    if "confidance" in attributes:
        attributes["confidence"] = float(attributes["confidance"])
        del attributes["confidance"]

display(annotations["annotations"])

new_annotations_file = data_dir / "annotations_cleaned.json"
with open(new_annotations_file, "w") as f:
    json.dump(annotations, f, indent=4)
```
